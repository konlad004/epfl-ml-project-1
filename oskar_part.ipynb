```python
import numpy as np
```

```python
from typing import Any
x_train = np.genfromtxt("Data/x_train.csv", delimiter=',', skip_header=1)
x_test = np.genfromtxt("Data/x_test.csv", delimiter=',', skip_header=1)
y_train: np.ndarray[tuple[int, ...], np.dtype[Any]] = np.genfromtxt("Data/y_train.csv", delimiter=',', skip_header=1)
```

```python
x_train_no_nan: np.ndarray[tuple[int, ...], np.dtype[Any]] = np.array([lis for lis in x_train if not (False in np.isnan(lis))])
```

```python
x_train_no_nan: np.ndarray[tuple[int, ...], np.dtype[Any]] = np.array([lis for lis in x_train.T if not (False in np.isnan(lis))])
```

```python
x_train_no_nan: np.ndarray[tuple[int, ...], np.dtype[Any]] = np.nan_to_num(x_train)
```

```python
y_train_no_nan: np.ndarray[tuple[int, ...], np.dtype[Any]] = np.nan_to_num(y_train)
```

```python
y_train_no_nan_res = y_train_no_nan[:,1]
```

```python
def least_squares(y, tx):

    w = np.linalg.solve(tx.T @ tx, tx.T @ y)
    return w, 1/(2*len(y))*(y- (tx @ w)).T @ (y- (tx @ w))
```

```python

def ridge_regression(y, tx, lambda_):

    N=len(y)
    w = np.linalg.solve(tx.T @ tx + (lambda_ * (2*N)) * np.identity(tx.shape[1]), tx.T @ y)
    return w, 1/(2*N)*(y- (tx @ w)).T @ (y- (tx @ w)) + lambda_ * w.T @ w
```

```python
w_ls, loss_ls = least_squares(y_train_no_nan_res, x_train_no_nan)
```

```python
w_rr, loss_r = ridge_regression(y_train_no_nan_res, x_train_no_nan, 2)
```
